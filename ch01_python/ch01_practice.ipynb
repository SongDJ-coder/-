## 1-5. 클래스
`Counter` 클래스
- 만들 때 `start` 값을 받아 `self.count`에 저장 (기본값 0)
- `add(n)`: count에 n을 더함
- `reset()`: count를 0으로

````python
class Counter:
    def __init__(self, start = 0):
        self.count = start

    def add(self, n):
        self.count += n
    
    def reset(self):
        self.count = 0
        


    
````

````python
# 확인
c = Counter(5)
c.add(3)
assert c.count == 8
c.reset()
assert c.count == 0 and Counter().count == 0
print('통과!')
````

## 1-7. 은행 계좌 클래스
`Account` 클래스를 만들자.
- 만들 때 `owner`(이름)와 `balance`(잔액, 기본값 0)를 받아 속성으로 저장
- `deposit(amount)`: 잔액에 amount를 더함
- `withdraw(amount)`: 잔액이 충분하면 빼고 `True` 반환, **부족하면 빼지 않고** `False` 반환
> 힌트: 메서드 첫 매개변수는 항상 `self`, 조건 분기는 `if`/`else`

````python
class Account:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return True

        else:
            return False
    
````

````python
t = Account('테스트', 1000)
print(t.withdraw(1000), t.balance)   # True 0  ← >= 로 고쳤다면
````

````python
# 확인
a = Account('송동준', 1000)
b = Account('친구')
a.deposit(500)
assert a.balance == 1500
assert a.withdraw(2000) is False and a.balance == 1500   # 잔액 부족 → 그대로
assert a.withdraw(300) is True and a.balance == 1200
assert b.owner == '친구' and b.balance == 0              # 기본값, a와 독립
print('통과!')
````

# 🧬 상속 (Inheritance)
이미 있는 클래스를 **물려받아서** 새 클래스를 만드는 것.

```python
class 자식클래스(부모클래스):
    ...
```
- 클래스 이름 뒤 괄호 = **부모 클래스** (1-5에서 `class Counter(start):`가 틀렸던 이유!)
- 자식은 부모의 메서드를 **그대로 쓸 수 있다** (다시 안 써도 됨)
- 같은 이름의 메서드를 자식에서 다시 만들면 **덮어쓴다 (오버라이드)**
- `super().메서드()` = 부모의 원래 메서드 호출

아래 예제를 실행해 보고 각 줄이 왜 그렇게 나오는지 생각해 보자.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return '...'

    def intro(self):
        return f'나는 {self.name}, ' + self.speak()


class Dog(Animal):              # Animal을 물려받음
    def speak(self):             # 오버라이드
        return '멍멍'


class Puppy(Dog):               # Dog을 물려받음 (Animal → Dog → Puppy)
    def __init__(self, name, age):
        super().__init__(name)   # 부모(Animal)의 __init__으로 name 저장
        self.age = age           # 새 속성 추가


a = Animal('동물')
d = Dog('바둑이')
p = Puppy('초코', 1)

print(a.intro())   # Animal의 speak
print(d.intro())   # intro는 Animal 것, speak는 Dog 것
print(p.intro(), p.age)
print(isinstance(p, Animal))   # Puppy도 Animal의 한 종류인가?
````

## 1-8. 상속 문제 ①: 메서드 추가
`Counter`를 물려받은 `StepCounter`를 만들자.
- 만들 때 `start`(기본 0)와 `step`(기본 1)을 받음 → `super().__init__(start)`로 count 설정, `self.step` 저장
- `tick()`: count를 step만큼 증가 (힌트: 물려받은 `add`를 쓰면 한 줄)
- `add`, `reset`은 **새로 만들지 않는다** (물려받은 걸 그대로 사용)

````python
class StepCounter(Counter):
    def __init__(self, start=0, step = 1): #여기서 궁금한 거는 부모 클래스에서 이미 stert = 으로 둔다는 규칙이 있으니까 자식 클래스에서는 추가할 필요 없는건가?
        super().__init__(start)
        self.step = step

    def tick(self):
        self.add(self.step)

## 기본 값이 없는게 뒤에 올 수 없다.
````

````python
# 확인
s = StepCounter(step=5)
s.tick(); s.tick()
assert s.count == 10
s.add(3)              # 물려받은 add
assert s.count == 13
s.reset()             # 물려받은 reset
assert s.count == 0 and s.step == 5
assert 'add' not in StepCounter.__dict__   # add를 다시 만들지 않았는지
print('통과!')
````

## 1-9. 상속 문제 ②: 오버라이드
`Counter`를 물려받은 `LimitCounter`를 만들자.
- 만들 때 `limit`(최댓값)을 받음, count는 0에서 시작
- `add(n)`을 **오버라이드**: 더한 결과가 limit을 넘으면 count를 limit으로 맞춤
- 힌트: `super().add(n)`으로 일단 더한 뒤, 넘쳤으면 고치기

````python
class LimitCounter(Counter):
    def __init__(self, limit):
        super().__init__()      ##부모 객체를 그냥 기져음. 
        self.limit = limit
    
    def add(self, n):
        super().add(n)    # 부모 클래스에 있는 add 를 호출해야 되므로 super()를 쓴다. 
        if self.count > self.limit:
            self.count = self.limit
        
````

````python
# 확인
m = LimitCounter(10)
m.add(4)
assert m.count == 4
m.add(100)
assert m.count == 10   # 10에서 멈춤
m.reset()
assert m.count == 0
print('통과!')
````

## 2-5. 원소별 연산 (2-1 복습)
`a = np.array([1, 2, 3, 4])`일 때
- `sq`: 각 원소의 제곱
- `centered`: 각 원소에서 평균(`a.mean()`)을 뺀 값
- `inv`: 각 원소의 역수 (1/원소)

````python
a = np.array([1, 2, 3, 4])
sq = a * a
centered = a - a.mean()
inv = np.array([1]) / a
# 브로드 캐스트의 규칙은 한 쪽이 1일 경우에 자동으로 늘린다.(같거나 한쪽이 1이면 늘린다. )
````

````python
# 확인
assert (sq == [1, 4, 9, 16]).all()
assert np.allclose(centered, [-1.5, -0.5, 0.5, 1.5])
assert np.allclose(inv, [1, 0.5, 1/3, 0.25])
print('통과!')
````
