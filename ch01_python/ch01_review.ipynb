## B-1. 클래스 (실수 3개)
`Timer`: `limit`은 꼭 받고, `start`는 기본값 0. `tick()`은 1씩 증가하되 limit을 넘지 않음.
> 고쳐야 할 것: 매개변수 순서, self, 경계값

````python
class Timer:
    def __init__(self, start=0, limit):
        self.time = start
        self.limit = limit

    def tick():
        if self.time <= self.limit:
            self.time += 1
````

````python
# 확인
t = Timer(3)
for _ in range(10):
    t.tick()
assert t.time == 3
assert Timer(5, start=4).time == 4
print('통과!')
````

## C-2. 브로드캐스트 예측 ⭐
**실행하지 말고** 결과 shape을 튜플로 적자. 에러가 나는 경우는 `'error'`라고 적는다.

| 계산 | 변수 |
|---|---|
| `(2,3)` + `(3,)` | `p1` |
| `(2,3)` + `(2,1)` | `p2` |
| `(2,3)` + `(2,)` | `p3` |
| `(4,1)` * `(3,)` | `p4` |
| `(3,)` / `(1,)` | `p5` |

> 규칙: 오른쪽 끝을 맞추고, 짧은 쪽 앞에 1을 붙이고, 각 자리가 **같거나 한쪽이 1**이면 OK.

````python
p1 = None  # 예: (2, 3)
p2 = None
p3 = None
p4 = None
p5 = None
````

````python
# 확인
def actual(a, b, op):
    try:
        return op(np.ones(a), np.ones(b)).shape
    except ValueError:
        return 'error'

import operator as o
cases = [((2,3),(3,),o.add,p1), ((2,3),(2,1),o.add,p2), ((2,3),(2,),o.add,p3),
         ((4,1),(3,),o.mul,p4), ((3,),(1,),o.truediv,p5)]
for i, (a, b, op, mine) in enumerate(cases, 1):
    ans = actual(a, b, op)
    print(f'p{i}: 내 답 {mine}  /  정답 {ans}  ' + ('✅' if mine == ans else '❌'))
assert all(actual(a, b, op) == mine for a, b, op, mine in cases)
print('통과!')
````
